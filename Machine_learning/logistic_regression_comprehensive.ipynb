```python
total_skus = 0
for fp in os.listdir(path):
    
    if fp == ".DS_Store":
        continue


    for f in os.listdir(os.path.join(path, fp)):
        if f == ".DS_Store":
            continue
        for f1 in os.listdir(os.path.join(path, fp, f, 'Excel_format')):
            if f1 == ".DS_Store":
                continue
            if f1.endswith(".xlsx"):
                

                print(f"Processing file: {f1}")
                df = pd.read_excel(os.path.join(path, fp, f, 'Excel_format', f1))
                print(df.shape)
                total_skus += df.shape[0]
            
        
```
