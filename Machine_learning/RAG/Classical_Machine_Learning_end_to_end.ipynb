```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Data Processing and ML Libraries
from sklearn.datasets import load_diabetes
from sklearn.model_selection import (
    train_test_split, 
    cross_val_score, 
    GridSearchCV, 
    StratifiedKFold, 
    KFold
)
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.feature_selection import SelectKBest, f_regression, RFE
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.metrics import (
    mean_squared_error, 
    mean_absolute_error, 
    r2_score, 
    explained_variance_score
)

# Models
from sklearn.linear_model import (
    LinearRegression, 
    Ridge, 
    Lasso, 
    ElasticNet
)
from sklearn.ensemble import (
    RandomForestRegressor, 
    GradientBoostingRegressor,
    VotingRegressor,
    StackingRegressor
)
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor

class AdvancedHousingDataProcessor:
    def __init__(self, test_size=0.2, random_state=42):
        # Load Boston Housing Dataset
        self.diabetes = load_diabetes()
        self.X = pd.DataFrame(self.diabetes.data, columns=self.diabetes.feature_names)
        self.y = pd.Series(self.diabetes.target)
        
        # Train-Test Split
        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(
            self.X, self.y, test_size=test_size, random_state=random_state
        )
        
        # Initialize results storage
        self.results = {}
    
    def advanced_feature_engineering(self):
        """Advanced Feature Engineering with Multiple Techniques"""
        # Polynomial Features
        poly = PolynomialFeatures(degree=2, include_bias=False)
        poly_features = poly.fit_transform(self.X_train)
        
        # # Interaction Features
        # self.X_train['LSTAT_CRIM'] = self.X_train['s1'] * self.X_train['s2']
        # self.X_train['RM_AGE'] = self.X_train['s3'] * self.X_train['s4']
        
        # # Log Transformation
        # log_features = ['s4', 's5']
        # for feature in log_features:
        #     self.X_train[f'log_{feature}'] = np.log1p(self.X_train[feature])
        
        return self.X_train
    
    def create_preprocessing_pipeline(self):
        """Create Comprehensive Preprocessing Pipeline"""
        numeric_features = self.X_train.columns
        
        preprocessor = ColumnTransformer(
            transformers=[
                ('num', Pipeline([
                    ('imputer', SimpleImputer(strategy='median')),
                    ('scaler', StandardScaler())
                ]), numeric_features)
            ])
        
        return preprocessor
    
    def cross_validation_evaluation(self, models, X, y):
        """Comprehensive Cross-Validation Evaluation"""
        cv_results = {}
        
        # Different Cross-Validation Strategies
        cv_strategies = {
            'KFold': KFold(n_splits=5, shuffle=True, random_state=42),
            # Add more CV strategies as needed
        }
        
        for name, model in models.items():
            cv_scores = {}
            
            for cv_name, cv in cv_strategies.items():
                # Multiple Scoring Metrics
                scoring_metrics = {
                    'neg_mean_squared_error': 'neg_mean_squared_error',
                    'r2': 'r2'
                }
                
                for metric_name, scoring in scoring_metrics.items():
                    scores = cross_val_score(
                        model, X, y, 
                        cv=cv, 
                        scoring=scoring
                    )
                    
                    cv_scores[f'{cv_name}_{metric_name}'] = {
                        'mean': scores.mean(),
                        'std': scores.std()
                    }
            
            cv_results[name] = cv_scores
        
        return cv_results
    
    def create_ensemble_models(self):
        """Create Advanced Ensemble Models"""
        # Base Models
        base_models = [
            ('rf', RandomForestRegressor(n_estimators=100, random_state=42)),
            ('gb', GradientBoostingRegressor(n_estimators=100, random_state=42)),
            ('svr', SVR(kernel='rbf'))
        ]
        
        # Voting Regressor
        voting_regressor = VotingRegressor(
            estimators=base_models
        )
        
        # Stacking Regressor
        stacking_regressor = StackingRegressor(
            estimators=base_models,
            final_estimator=LinearRegression(),
            cv=5
        )
        
        return {
            'Voting Regressor': voting_regressor,
            'Stacking Regressor': stacking_regressor
        }
    
    def hyperparameter_tuning(self, models):
        """Advanced Hyperparameter Tuning"""
        tuned_models = {}
        
        # Hyperparameter Grids
        param_grids = {
            'Random Forest': {
                'n_estimators': [50, 100, 200],
                'max_depth': [None, 10, 20, 30],
                'min_samples_split': [2, 5, 10]
            },
            'Gradient Boosting': {
                'n_estimators': [50, 100, 200],
                'learning_rate': [0.01, 0.1, 0.5],
                'max_depth': [3, 5, 7]
            }
        }
        
        for name, model in models.items():
            if name in param_grids:
                grid_search = GridSearchCV(
                    model, 
                    param_grids[name], 
                    cv=5, 
                    scoring='neg_mean_squared_error',
                    n_jobs=-1
                )
                grid_search.fit(self.X_train, self.y_train)
                
                tuned_models[name] = {
                    'best_model': grid_search.best_estimator_,
                    'best_params': grid_search.best_params_
                }
            else:
                tuned_models[name] = {'best_model': model}
        
        return tuned_models
    
    def final_model_evaluation(self, tuned_models):
        """Comprehensive Model Evaluation"""
        evaluation_results = {}
        
        for name, model_info in tuned_models.items():
            model = model_info['best_model']
            
            # Fit the model
            model.fit(self.X_train, self.y_train)
            
            # Predictions
            y_pred = model.predict(self.X_test)
            
            # Evaluation Metrics
            evaluation_results[name] = {
                'MSE': mean_squared_error(self.y_test, y_pred),
                'MAE': mean_absolute_error(self.y_test, y_pred),
                'R2': r2_score(self.y_test, y_pred),
                'Best Params': model_info.get('best_params', {})
            }
        
        return evaluation_results
    
    def run_complete_workflow(self):
        """End-to-End Machine Learning Workflow"""
        # Feature Engineering
        X_engineered = self.advanced_feature_engineering()
        
        # Preprocessing Pipeline
        preprocessor = self.create_preprocessing_pipeline()
        
        # Base Models
        base_models = {
            'Random Forest': RandomForestRegressor(random_state=42),
            'Gradient Boosting': GradientBoostingRegressor(random_state=42),
            'SVR': SVR(),
            'Linear Regression': LinearRegression()
        }
        
        # Cross-Validation
        cv_results = self.cross_validation_evaluation(base_models, X_engineered, self.y_train)
        print("Cross-Validation Results:")
        for model, results in cv_results.items():
            print(f"\n{model}:")
            for metric, score in results.items():
                print(f"{metric}: {score}")
        
        # Ensemble Models
        ensemble_models = self.create_ensemble_models()
        base_models.update(ensemble_models)
        
        # Hyperparameter Tuning
        tuned_models = self.hyperparameter_tuning(base_models)
        
        # Final Model Evaluation
        final_results = self.final_model_evaluation(tuned_models)
        
        print("\nFinal Model Evaluation:")
        for model, metrics in final_results.items():
            print(f"\n{model}:")
            for metric, value in metrics.items():
                print(f"{metric}: {value}")
        
        return final_results

def main():
    # Initialize and Run Workflow
    processor = AdvancedHousingDataProcessor()
    final_results = processor.run_complete_workflow()


main()
```
