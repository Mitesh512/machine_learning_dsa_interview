# Introduction to Logistic Regression

## Overview
Logistic regression is a fundamental classification algorithm used in machine learning and statistics. Despite its name, it's used for binary classification problems rather than regression.

## Key Concepts
1. Binary Classification
2. Sigmoid Function
3. Decision Boundary
4. Probability Estimation

Let's explore each concept with theory and implementation.

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_classification

# Set random seed for reproducibility
np.random.seed(42)
```

## 1. The Sigmoid Function

The sigmoid function (also called logistic function) is the heart of logistic regression. It maps any real number to the range [0,1].

Mathematical form:
$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

Let's visualize this function:

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Generate points for visualization
z = np.linspace(-10, 10, 100)
sigmoid_values = sigmoid(z)

plt.figure(figsize=(10, 6))
plt.plot(z, sigmoid_values)
plt.title('Sigmoid Function')
plt.xlabel('z')
plt.ylabel('σ(z)')
plt.grid(True)
plt.show()
```

## 2. Simple Binary Classification Example

Let's create a simple dataset and visualize how logistic regression works:

```python
# Generate synthetic dataset
X, y = make_classification(
    n_samples=100,
    n_features=2,
    n_redundant=0,
    n_informative=2,
    random_state=42,
    n_clusters_per_class=1
)

# Visualize the data
plt.figure(figsize=(10, 6))
plt.scatter(X[y == 0][:, 0], X[y == 0][:, 1], label='Class 0')
plt.scatter(X[y == 1][:, 0], X[y == 1][:, 1], label='Class 1')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Binary Classification Dataset')
plt.legend()
plt.show()
```

## 3. Mathematical Foundation

In logistic regression, we model the probability of the positive class (y=1) given features X:

$$P(y=1|X) = \sigma(w^TX + b)$$

where:
- w is the weight vector
- b is the bias term
- X is the input feature vector

The decision rule is:
- If P(y=1|X) ≥ 0.5, predict class 1
- If P(y=1|X) < 0.5, predict class 0

```python
class SimpleLogisticRegression:
    def __init__(self, learning_rate=0.01, n_iterations=1000):
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.weights = None
        self.bias = None
    
    def fit(self, X, y):
        n_samples, n_features = X.shape
        
        # Initialize parameters
        self.weights = np.zeros(n_features)
        self.bias = 0
        
        # Gradient descent
        for _ in range(self.n_iterations):
            # Forward pass
            linear_model = np.dot(X, self.weights) + self.bias
            y_predicted = sigmoid(linear_model)
            
            # Compute gradients
            dw = (1/n_samples) * np.dot(X.T, (y_predicted - y))
            db = (1/n_samples) * np.sum(y_predicted - y)
            
            # Update parameters
            self.weights -= self.learning_rate * dw
            self.bias -= self.learning_rate * db
    
    def predict(self, X):
        linear_model = np.dot(X, self.weights) + self.bias
        y_predicted = sigmoid(linear_model)
        return y_predicted >= 0.5
```

Let's train our simple logistic regression model and visualize the decision boundary:

```python
# Train the model
model = SimpleLogisticRegression(learning_rate=0.1, n_iterations=1000)
model.fit(X, y)

# Create a mesh grid to visualize the decision boundary
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1),
                     np.arange(y_min, y_max, 0.1))

# Make predictions on the mesh grid
Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Plot the decision boundary
plt.figure(figsize=(10, 6))
plt.contourf(xx, yy, Z, alpha=0.4)
plt.scatter(X[y == 0][:, 0], X[y == 0][:, 1], label='Class 0')
plt.scatter(X[y == 1][:, 0], X[y == 1][:, 1], label='Class 1')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Decision Boundary')
plt.legend()
plt.show()
```
