```python
import shap
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier

# Load data
X, y = load_iris(return_X_y=True)

# Train model
model = RandomForestClassifier().fit(X, y)

# Create SHAP explainer (TreeExplainer for tree models)
explainer = shap.TreeExplainer(model)

# Explain predictions for first 5 samples
shap_values = explainer.shap_values(X[:5])

print(shap_values)
# # Plot explanation for first sample, class 0
print(explainer.expected_value[0].shape, shap_values[0][0].shape, X[0].shape)
# shap.initjs()
# shap.force_plot(explainer.expected_value[0], shap_values[0][0], X[0])
```
