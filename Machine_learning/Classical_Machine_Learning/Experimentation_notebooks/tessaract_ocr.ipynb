```python
# Load the image
image_path = "pan_sample.jpeg"#"hand_written_text2.jpeg" #"hand_written_sample.jpeg"#"pan_sample.jpeg"  # Replace with your image path
image = cv2.imread(image_path)
```

```python
# Step 1: Preprocessing
# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
# Step 2: Use Tesseract to extract text
text = pytesseract.image_to_string(gray)

# Step 3: Apply post-processing
```

```python
# Use Tesseract with specific configurations for better accuracy
custom_config = r'--oem 3 --psm 4'  # OEM 3: Default, PSM 6: Assume a single uniform block of text
corrected_text = pytesseract.image_to_string(gray, config=custom_config)

# Print the corrected text
print("Corrected Text (Tesseract):")
print(corrected_text)
```

```python
import re

# Function to extract PAN number
def extract_pan(text):
    # PAN card pattern: 5 alphabets, followed by 4 numbers, followed by 1 alphabet
    pan_pattern = r'-[A-Z]{10}' #r'[A-Z]{5}[0-9]{4}[A-Z]{1}'
    
    # Search for PAN pattern in text
    match = re.search(pan_pattern, text)
    
    if match:
        return match.group()
    return None

# Extract PAN from the corrected text
pan_number = extract_pan(corrected_text)
if pan_number:
    print(f"{pan_number}")
else:
    print("No valid PAN number found")
```
