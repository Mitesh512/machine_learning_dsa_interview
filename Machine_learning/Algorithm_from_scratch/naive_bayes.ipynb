# What is Baye's Theorem why is it needed?



Bayes' Theorem helps us update our beliefs about an event based on new evidence.

Analogy:
Imagine you’re a doctor diagnosing a rare disease (D) based on a positive test result (T). The test isn’t perfect—it sometimes gives false positives. Bayes' Theorem helps you calculate the probability of the patient actually having the disease given a positive test result, by combining:

- How likely the disease is in the population (prior belief).
- How accurate the test is (likelihood of evidence).
- The overall probability of a positive test result.

P(A∣B): Posterior probability — the probability of hypothesis A given the data B.

P(B∣A): Likelihood — the probability of data B given that hypothesis A is true.

P(A): Prior — the probability of hypothesis A before seeing the data.

P(B): Evidence — the total probability of the data under all possible hypotheses.

Applications:


Foundation for Probabilistic Models: Algorithms like Naive Bayes, Bayesian Networks, and Gaussian Processes rely on it.
Naive Bayes Classifier:
Used for classification tasks (e.g., spam detection, sentiment analysis).
Assumes features are conditionally independent given the class.

# Naive Bayes Classifier Code:

Text Classification : Spam vs Ham (not spam)


```python
import math
from collections import defaultdict


class NaiveBayesClassifier:
    def __init__(self):
        # count of each word per class class: word: frequency
        self.class_word_count = defaultdict(lambda: defaultdict(int))
        
        # total documents per class
        self.class_count = defaultdict(int)
        
        # Total number of documents
        self.total_docs = 0
        
        # vocabulary to keep track of total number of words
        self.vocab = set()
        

    def tokenize(self,text):
        # convert to lowercase words
        return text.lower().split()
    
    
    
    def fit(self,X,y):
        """
        We will train the Naive Bayes Classifier
        """
        for text,label in zip(X,y):
            self.total_docs += 1
            self.class_count[label] += 1
            
            # tokenize the document and update word frequencies
            for word in self.tokenize(text):
                self.class_word_count[label][word] += 1
                self.vocab.add(word)        

    
    def predict(self,text):
        words = self.tokenize(text)
        
        class_scores = {}
        
        for cls in self.class_count:
            # start with the log of proabilities of the prior probabilities
            # how many belog to a class 
            log_prob = math.log(self.class_count[cls]/self.total_docs)
            
            # total word count in this class
            total_words_in_class = sum(self.class_word_count[cls].values())
            # print(total_words_in_class)
            
            # total_vocab_size 
            total_vocab_size = len(self.vocab)
            
            
            for word in words:
                # first get the wrod frequency from training data
                word_freq = self.class_word_count[cls][word] + 1 # +1 for laplace smoothing
                
                # word_probability
                word_prob = word_freq/(total_words_in_class + total_vocab_size)
                
                # log_probability 
                log_prob += math.log(word_prob)
            
            class_scores[cls] = log_prob
            
        
        return max(class_scores, key = class_scores.get)


                
                
        
    
    
```

```python
# Sample training dataset
X_train = [
    "Win money now",
    "Limited offer just for you",
    "Call me later",
    "Let's have lunch tomorrow",
    "Congratulations you won a lottery",
    "Are we still meeting today"
]

# Labels: spam vs ham (not spam)
y_train = ["spam", "spam", "ham", "ham", "spam", "ham"]

# Test examples we want to classify
X_test = [
    "win a lottery now",
    "let's meet for lunch",
    "exclusive offer win money"
]

# Step 1: Create and train the classifier
nb = NaiveBayesClassifier()
nb.fit(X_train, y_train)


# Step 2: Predict on test data
for text in X_test:
    pred = nb.predict(text)
    print(f"Text: '{text}' => Prediction: {pred}")


    
```
