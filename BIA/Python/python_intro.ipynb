```python
def calculate_tax(income):
    if income > 50000:
        tax_rate = 0.20
    else:
        tax_rate = 0.15
    tax_amount = income * tax_rate
    return tax_amount
income = float(input("Enter your income: $"))
tax = calculate_tax(income)
print(f"Your tax amount is: ${tax:.2f}")
```

```python
def calculate_savings(monthly_deposit, annual_interest_rate, months):
    savings = 0.0
    monthly_interest_rate = annual_interest_rate / 12 / 100

    for month in range( months ):
        savings += monthly_deposit  # Deposit first
        savings += savings * monthly_interest_rate  # Add interest
    return savings
```

```python
def calculate_debt_repayment(initial_debt, monthly_payment, annual_interest_rate):
    debt = initial_debt
    months = 0
    monthly_interest_rate = annual_interest_rate / 12 / 100

    while debt > 0:
        debt += debt * monthly_interest_rate  # Add interest
        debt -= monthly_payment              # Make payment
        months += 1

        if debt < 0:  # Optional: Avoid negative final balance
            debt = 0

    return months
```

```python
def loan_amortization_schedule(loan_amount, annual_rate, years):
    monthly_rate = annual_rate / 12 / 100
    num_payments = years * 12

    # Monthly payment formula
    monthly_payment = loan_amount * monthly_rate / (1 - (1 + monthly_rate) ** -num_payments)

    print(f"\nMonthly Payment: ${monthly_payment:.2f}")
    print("Amortization Schedule:")
    print(f"{'Month':<6}{'Payment':<10}{'Interest':<10}{'Principal':<12}{'Balance'}")

    balance = loan_amount
    for month in range(1, num_payments + 1):
        interest = balance * monthly_rate
        principal = monthly_payment - interest
        balance -= principal
        if balance < 0:
            balance = 0
        print(f"{month:<6}{monthly_payment:<10.2f}{interest:<10.2f}{principal:<12.2f}{balance:.2f}")

# Example usage:
loan_amortization_schedule(100000, 5, 15)
```

```python
def calculate_future_value(initial_investment, annual_rate, years):
    future_value = initial_investment * ((1 + annual_rate / 100) ** years)
    return future_value
```

```python
def analyze_budget(budget, expenses):
    total_expenses = sum(expenses)
    difference = budget - total_expenses

    if difference > 0:
        status = "within"
    elif difference < 0:
        status = "over"
    else:
        status = "exactly on"

    summary = (
        f"Total Budget: ${budget}\n"
        f"Total Expenses: ${total_expenses}\n"
        f"You are {status} your budget by ${abs(difference):.2f}."
    )
    return summary
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.pairplot(bank_data[['CreditScore', 'Age', 'Balance', 'NumOfProducts', 'Exited']], hue='Exited', markers=['o', 's'])
plt.suptitle('Pairplot of Selected Features by Churn Status', y=1.02)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=bank_data, x='Age', hue='Exited', multiple='stack', kde=True, palette='viridis')
plt.title('Age Distribution by Churn Status')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
correlation_matrix = bank_data.corr(numeric_only=True)
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Matrix - Bank Churn Data')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x='Gender', hue='Exited', data=bank_data, palette='pastel')
plt.title('Churn by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x='Exited', y='Balance', data=bank_data, palette='Set2')
plt.title('Balance Distribution by Churn Status')
plt.xlabel('Churn Status')
plt.ylabel('Balance')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.violinplot(x='Exited', y='Age', data=bank_data, palette='Set3')
plt.title('Age Distribution by Churn Status')
plt.xlabel('Churn Status')
plt.ylabel('Age')
plt.show()
```
